```python
n = 5
symbol = "#"


def extra_spacing(current, end, pattern, output):

    if current != end:
        output += pattern + " "
    else:
        output += pattern

    return output
```

```python
def numerical_square_pattern(n=4, pattern="0"):

    for a in range(1, n + 1):

        output = ""

        for b in range(1, n + 1):
            output = extra_spacing(b, n, pattern, output)

        print(output)

    return None


numerical_square_pattern(n, "1")
```

```python
def numerical_incremental_pattern(n=4, pattern="1"):

    for a in range(1, n + 1):

        output = ""

        for b in range(1, n + 1):
            output = extra_spacing(b, n + 1, f"{a}", output)

        print(output)

    return None


numerical_incremental_pattern(n)
```

```python
def internal_varsity(n=4):

    pattern = f"{n}"

    # starting pattern
    output = ""

    for x in range(1, n + 1):
        output = extra_spacing(x, n, pattern, output)

    print(output)

    # internal varsity pattern
    counter = 1

    for a in range(1, n):

        output = ""
        for b in range(1, n + 1):

            if b == 1 or b == n:
                output = extra_spacing(b, n, pattern, output)
            else:
                output = extra_spacing(b, n, f"{counter}", output)

        print(output)
        counter += 1

    # ending pattern
    output = ""
    for y in range(1, n + 1):
        output = extra_spacing(y, n, pattern, output)

    print(output)

    return None


internal_varsity()
```

```python
def inverted_right_angle_triangle_pattern(n=4):

    pattern = n

    for a in range(1, n + 2):

        output = ""

        for b in range(1, a + 1):
            output = extra_spacing(b, a, f"{pattern}", output)

        print(output)
        pattern += 1

    return None


inverted_right_angle_triangle_pattern(n)
```

```python
def double_incrementing_pattern(n=4):

    pattern = n

    for a in range(1, n + 2):

        output = ""

        for b in range(1, a + 1):
            output = extra_spacing(b, a, f"{pattern}", output)
            pattern += 1

        print(output)

    # resetting the pattern value to current loop's last number value and assigning it to end var
    pattern -= 1
    pattern -= n
    end = pattern

    for c in range(0, n):

        # setting the pattern to the start number value of the current loop
        pattern -= n - c
        output = ""

        for d in range(pattern, end):
            output = extra_spacing(d, end - 1, f"{d}", output)

        print(output)
        end = pattern

    return None


double_incrementing_pattern(n)
```

```python
def inverted_right_angle_triangle_pattern(n=4):

    start = 0

    # loop to get the starting number (pattern)
    for x in range(1, n + 1):
        start += x

    for a in range(1, n + 1):

        output = ""

        for b in range(1, a + 1):
            output = extra_spacing(b, a, f"{start}", output)
            start -= 1

        print(output)

    return None


inverted_right_angle_triangle_pattern(5)
```

```python
def inverted_diamond_pattern(n=4):

    pattern = n

    for a in range(1, n + 2):

        output = ""
        for b in range(1, a + 1):
            output = extra_spacing(b, a, f"{pattern}", output)

        pattern += 1
        print(output)

    pattern -= 2

    for c in range(n, 0, -1):

        output = ""
        for d in range(1, c + 1):
            output = extra_spacing(d, c, f"{pattern}", output)

        print(output)
        pattern -= 1

    return None


inverted_diamond_pattern()
```

```python
def inverted_diamond_pattern(n=4):

    pattern = n - 1

    for a in range(1, 5):

        output = ""

        for b in range(1, a + 1):
            output = extra_spacing(b, a, f"{pattern}", output)

        print(output)
        pattern += 1

    pattern -= 1

    for c in range(4, 0, -1):

        output = ""

        for b in range(1, c + 1):
            output = extra_spacing(b, c, f"{pattern}", output)

        print(output)
        pattern -= 1

    return None


inverted_diamond_pattern(n)
```

**Largest Element**\
Given an array of integers nums, return the value of the largest element in the array 

```typescript
function largestElement(array: number[]) {

    let n = array.length;
    let max = array[0];

    for (let a = 1; a < n; a++) {

        let current = array[a];
        if (current > max) {
            max = current;
        }
    }

    return max;
}

let output = largestElement([3, 3, 0, 99, -40])
console.log(output)
```

**Find Second Smallest and Second Largest Element in an array**\
Given an array, find the second smallest and second largest element in the array. Print ‘-1’ in the event that either of them doesn’t exist.

```typescript
function secondSmallestAndLargest(array: number[]): number[] | -1 {

    let output: number[] = [];
    let n = array.length;

    if (n < 4) {
        return -1
    }


    let smallest = Infinity;
    let secondSmallest = Infinity

    for (let a = 0; a < n; a++) {

        let current = array[a];

        if (current < smallest) {
            secondSmallest = smallest;
            smallest = current;
        } else if (current < secondSmallest && current !== smallest) {
            secondSmallest = current;
        }


    }

    let largest = -Infinity;
    let secondLargest = -Infinity;

    for (let a = 0; a < n; a++) {

        let current = array[a];

        if (current > largest) {
            secondLargest = largest;
            largest = current;
        } else if (current > secondLargest && current !== largest) {
            secondLargest = current;
        }

    }

    output.push(...[secondSmallest, secondLargest])

    return output
}


let outputOne = secondSmallestAndLargest([122, 113, 234, 647]);
console.log(outputOne)
```

**Check if an Array is Sorted**\
Given an array of size n, write a program to check if the given array is sorted in (ascending / Increasing / Non-decreasing) order or not. If the array is sorted then return True, Else return False.

```typescript
function isArraySorterd(array: number[]): boolean {

    let n = array.length

    for (let a = 1; a < n; a++) {

        let current = array[a];
        let previous = array[a - 1];

        if (current < previous) {
            return false;
        }

    }

    return true;
}


let output = isArraySorterd([1, 2, 4, 5]);
console.log(output)
```

**Remove Duplicates in-place from Sorted Array**\
Given an integer array sorted in non-decreasing order, remove the duplicates in place such that each unique element appears only once. The relative order of the elements should be kept the same.

If there are k elements after removing the duplicates, then the first k elements of the array should hold the final result. It does not matter what you leave beyond the first k elements.

```typescript
function removeDuplicates(array: number[]): [number, number[]] {

    let n = array.length;

    if (n < 1) { return array }

    let left = 0;
    let right = 1;
    let count = 1;

    while (right < n && left < right) {

        let positional = array[left];
        let current = array[right];

        if (current !== positional && right !== left + 1) {
            left++;
            let ref = positional;
            array[left] = current;
            array[right] = ref;

            right++;
            count++;
        } else if (current === positional) {
            right++;
        } else {
            left++;
            right++;
            count++;
        }

    }
    return [count, array];
}

let output = removeDuplicates([1, 1, 2, 3, 3, 4, 5, 5, 5, 6]);
console.log(output);
```

**Left Rotate the Array by One**\
Given an integer array nums, rotate the array to the left by one.

Note: There is no need to return anything, just modify the given array.

```typescript
function leftRotateArrayOne(array: number[]) {

    let n = array.length;

    for (let a = 1; a < n; a++) {

        let previous = array[a - 1];
        let current = array[a];

        array[a - 1] = current;
        array[a] = previous;

    }

    console.log(array);
}

leftRotateArrayOne([1, 2, 3, 4, 5]);
```

```typescript
const n = 5;
const symbol = "^";


function extraSpacing(current: number, end: number, pattern: string, output: string) {

    if (current !== end) {
        output += pattern + " ";
    } else {
        output += pattern;
    }

    return output
}
```

```typescript
function numericalTwosComplementPattern(n = 4) {

    let pattern = 1
    for (let a = 1; a <= n; a++) {

        let output = "";
        let localPattern = pattern * 2;

        for (let b = 1; b <= a; b++) {
            output = extraSpacing(b, a, `${b === 1 ? pattern : localPattern}`, output)
        }

        console.log(output)
        pattern *= 2
    }
}
numericalTwosComplementPattern(n)
```

```typescript
function numberReductionPattern(n = 4) {


    for (let a = 1; a <= n; a++) {

        let output = ""

        for (let b = a; b <= n; b++) {
            output = extraSpacing(b, n, `${b}`, output);
        }

        console.log(output)
    }
}

numberReductionPattern(n)
```

```typescript
function doubleColumnValuePattern(n = 4) {


    for (let a = 0; a <= n; a++) {

        let output = "";

        for (let b = 0; b <= a; b++) {

            output = extraSpacing(b, a, `${a * b}`, output);
        }

        console.log(output)
    }

}

doubleColumnValuePattern(n)
```

```typescript
function uniquePyramidPattern(n = 4) {

    for (let a = 1; a <= n; a++) {

        let output = "";
        let pattern = 1

        for (let b = 1; b <= (2 * a) - 1; b++) {
            output = extraSpacing(b, (2 * a) - 1, `${pattern}`, output);

            if (b < a) {
                pattern++;
            } else {
                pattern--;
            }
        }

        console.log(output);
    }



}
uniquePyramidPattern(n)
```

```typescript
function evenNumberPattern(n = 4) {

    for (let a = 1; a <= n; a++) {

        let output = "";
        let pattern = 2 * n;

        for (let b = 1; b <= a; b++) {

            output = extraSpacing(b, a, `${pattern}`, output);
            pattern -= 2;
        }

        console.log(output);
    }
}

evenNumberPattern(n)
```

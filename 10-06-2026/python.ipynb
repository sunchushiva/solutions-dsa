```python
n = 5
symbol = "^"


def extra_spacing(current, end, pattern, output):

    if current != end:
        output += pattern + " "
    else:
        output += pattern

    return output
```

```python
def double_the_number_pattern(n=4):

    for a in range(0, n + 1):

        output = ""
        pattern = 2**a

        for b in range(0, a + 1):
            output = extra_spacing(b, a, f"{pattern}", output)
            pattern = pattern // 2

        print(output)

    return None


double_the_number_pattern(n)
```

```python
def equilateral_triangle_pattern(n=4):

    pattern = 65
    space = n - 1
    for a in range(1, n + 1):

        output = ""

        for x in range(1, space + 1):
            output = extra_spacing(x, space + 1, " ", output)

        for b in range(1, (2 * a)):

            if pattern > 90:
                pattern = 65

            output = extra_spacing(b, (2 * a) - 1, chr(pattern), output)
            pattern += 1

        print(output)
        space -= 1

    return None


equilateral_triangle_pattern(n)
```

```python
def hollow_diamond_pattern(n=4, pattern="*"):

    space = n - 1

    # top half pattern
    for a in range(1, n + 1):

        output = ""

        # initial spacing loop
        for x in range(1, space + 1):
            output = extra_spacing(x, space + 1, " ", output)

        # pattern printing loop
        for b in range(1, (2 * a)):
            if b == 1 or b == (2 * a) - 1:
                output = extra_spacing(b, (2 * a) - 1, f"{pattern}", output)
            else:
                output = extra_spacing(b, (2 * a) - 1, " ", output)

        print(output)
        space -= 1

    space = 1

    # bottom half pattern
    for a in range(n - 1, 0, -1):

        output = ""

        # initial spacing loop
        for x in range(1, space + 1):
            output = extra_spacing(x, space + 1, " ", output)

        # pattern printing loop
        for b in range(1, (2 * a)):
            if b == 1 or b == (2 * a) - 1:
                output = extra_spacing(b, (2 * a) - 1, f"{pattern}", output)
            else:
                output = extra_spacing(b, (2 * a) - 1, " ", output)

        print(output)
        space += 1

    return None


hollow_diamond_pattern(n, symbol)
```

```python
def numerical_square_pattern(n=4):

    for a in range(1, n + 1):

        output = ""

        for b in range(1, n + 1):
            output = extra_spacing(b, n, f"{b if b>a else a}", output)

        print(output)

    return None


numerical_square_pattern(n)
```

```python
def inverted_numerical_triangle_pattern(n=4):

    space = n - 1

    for a in range(1, n + 1):

        output = ""

        for x in range(1, space + 1):
            output = extra_spacing(x, space + 1, " ", output)

        for b in range(1, a + 1):
            output = extra_spacing(b, a, f"{b}", output)

        print(output)
        space -= 1

    return None


inverted_numerical_triangle_pattern(n)
```

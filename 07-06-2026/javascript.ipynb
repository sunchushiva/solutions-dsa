```typescript
const n = 5;
const symbol = "&";

function extraSpacing(current: number, end: number, pattern: string, output: string) {

    if (current !== end) {
        output += pattern + "!";
    } else {
        output += pattern;
    }
    return output;
}
```

```typescript
function invertedSquaredNumericalPattern(n = 4) {

    // values for the current loop's starting and ending number
    let end = n ** 2;
    let start = end - n;

    for (let a = 1; a <= n; a++) {

        let output = "";

        for (let b = start + 1; b <= end; b++) {

            output = extraSpacing(b, end, `${b}`, output)
        }

        console.log(output)
        start -= n;
        end -= n;
    }

}

invertedSquaredNumericalPattern(n)
```

```typescript
function incrementalSquaredNumericalAlternatePattern(n = 4) {

    let odd = 1;
    let even = (n * 2) + 1

    for (let a = 1; a <= n; a++) {

        let output = "";
        for (let b = 1; b <= n; b++) {
            output = extraSpacing(b, n, `${a % 2 === 0 ? even : odd}`, output);
            if (a % 2 === 0) {
                even++;
            } else {
                odd++;
            }
        }

        console.log(output)
    }

}
incrementalSquaredNumericalAlternatePattern(n)
```

```typescript
function incrementalSandwhichNumericalAlternatePattern(n = 4) {

    let outlier = 1;
    let normal = (n * 2) + 1

    for (let a = 1; a <= n; a++) {

        let output = "";

        for (let b = 1; b <= n; b++) {

            output = extraSpacing(b, n, `${a === 1 || a === n ? outlier : normal}`, output)

            if (a === 1 || a === n) {
                outlier++;
            } else {
                normal++;
            }

        }

        console.log(output)
    }

}
incrementalSandwhichNumericalAlternatePattern(n)
```

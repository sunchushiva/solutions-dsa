```typescript
const n = 5;
const symbol = "#";


function extraSpacing(current: number, end: number, pattern: string, output: string) {

    if (current !== end) {
        output += pattern + " ";
    } else {
        output += pattern;
    }

    return output;
}
```

```typescript
function halfDiamondPattern(n = 4, pattern = "*") {

    // First half pattern 

    for (let a = 1; a <= n; a++) {

        let output = "";

        for (let b = 1; b <= a; b++) {
            output = extraSpacing(b, a, pattern, output);
        }

        console.log(output);

    }

    // Second half pattern

    for (let c = n - 1; c >= 1; c--) {

        let output = "";

        for (let d = 1; d <= c; d++) {
            output = extraSpacing(d, c, pattern, output);
        }

        console.log(output);
    }

}

halfDiamondPattern(n, symbol);
```

```typescript
function invertedHalfDiamondPattern(n = 4, pattern = "*") {

    let spacing = n - 1;

    // First half pattern 
    for (let a = 1; a <= n; a++) {

        let output = "";

        // initial spacing loop
        for (let x = 1; x <= spacing; x++) {
            output = extraSpacing(x, spacing + 1, " ", output);
        }

        for (let b = 1; b <= a; b++) {
            output = extraSpacing(b, a, pattern, output);
        }

        console.log(output);
        spacing--;

    }

    // Second half pattern 
    spacing = 1;

    for (let c = n - 1; c >= 1; c--) {

        let output = "";

        // initial spacing loop
        for (let y = 1; y <= spacing; y++) {
            output = extraSpacing(y, spacing + 1, " ", output);
        }


        for (let d = 1; d <= c; d++) {
            output = extraSpacing(d, c, pattern, output);
        }

        console.log(output);
        spacing++;

    }
}

invertedHalfDiamondPattern(n, symbol);
```

```typescript
function diamondPattern(n = 4, pattern = "*") {

    let stars = 1;
    let spacing = n - 1;

    // First half pattern 
    for (let a = 1; a <= n; a++) {

        let output = "";

        // initial spacing loop
        for (let x = 1; x <= spacing; x++) {
            output = extraSpacing(x, spacing + 1, " ", output);
        }

        for (let b = 1; b <= stars; b++) {
            output = extraSpacing(b, stars, pattern, output);
        }

        console.log(output);
        stars += 2;
        spacing--;
    }


    stars -= 4;
    spacing += 2

    // Second half pattern
    for (let c = 1; c <= n - 1; c++) {

        let output = "";

        // initial spacing loop
        for (let y = 1; y <= spacing; y++) {
            output = extraSpacing(y, spacing + 1, " ", output);
        }

        for (let d = 1; d <= stars; d++) {
            output = extraSpacing(d, stars, pattern, output);
        }

        console.log(output);
        stars -= 2;
        spacing += 1;
    }

}

diamondPattern(n, symbol);
```

```typescript
function rightAngleTrianglePattern(n = 4) {

    let pattern = 1;
    let columns = 1;

    while (columns <= n) {

        let output = "";

        for (let a = 1; a <= columns; a++) {
            output = extraSpacing(a, columns, `${pattern}`, output);
            pattern++;
        }

        console.log(output);
        columns++;

    }
}

rightAngleTrianglePattern(n);
```

```typescript
function incrementingTrianglePattern(n = 4) {

    let pattern = 2 * n;

    for (let a = n + 1; a >= 1; a--) {

        let output = "";

        for (let b = 1; b <= a; b++) {

            output = extraSpacing(b, a, `${pattern}`, output);

        }

        console.log(output);
        pattern--;

    }


}

incrementingTrianglePattern(n);
```

```typescript
function doubleIncrementingTrianglePattern(n = 4) {

    let pattern = 3;
    let columns = 1;

    for (let b = 1; b <= n; b++) {

        let output = "";

        for (let a = 1; a <= columns; a++) {
            output = extraSpacing(a, columns, `${pattern}`, output);
            pattern++;
        }

        columns++;
        console.log(output);
    }
}

doubleIncrementingTrianglePattern(n);
```

```typescript
function numericalSquaredPattern(n = 4) {

    let pattern = 1;

    for (let a = 1; a <= n; a++) {

        let output = "";

        for (let b = 1; b <= n; b++) {

            output = extraSpacing(b, n, `${pattern}`, output);
            pattern++;
        }

        console.log(output);
    }
}

numericalSquaredPattern(5);
```

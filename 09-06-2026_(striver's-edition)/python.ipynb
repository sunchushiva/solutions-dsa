```python
n = 5
symbol = "%"


def extra_spacing(current, end, pattern, output):

    if current != end:
        output += pattern + " "
    else:
        output += pattern

    return output
```

```python
def inverse_alphabetical_pattern(n=4):

    for a in range(n, 0, -1):

        pattern = 65
        output = ""

        for b in range(1, a + 1):

            if pattern > 90:
                pattern = 65

            output = extra_spacing(b, a, chr(pattern), output)
            pattern += 1

        print(output)

    return None


inverse_alphabetical_pattern(n)
```

```python
def inverse_alphabetical_pattern_two(n=4):

    ref = n % 26
    pattern = 64 + (ref if ref > 0 else 1)

    for a in range(1, n + 1):

        output = ""

        if pattern < 65:
            pattern = 90

        # variable for control values column values in current row
        local_pattern = pattern

        for b in range(1, a + 1):

            if local_pattern > 90:
                local_pattern = 65

            output = extra_spacing(b, a, chr(local_pattern), output)
            local_pattern += 1

        print(output)
        pattern -= 1

    return None


inverse_alphabetical_pattern_two(n)
```

```python
def inverted_numerical_pyramid_pattern(n=4):

    spaces = (n * 2) - 1

    for a in range(1, n + 1):

        output = ""

        # loop for first half pattern
        for b in range(1, a + 1):
            output = extra_spacing(b, a, f"{b}", output)

        # loop for spacing
        for x in range(1, spaces + 1):
            output = extra_spacing(x, spaces, " ", output)

        # loop for second half pattern
        for c in range(a, 0, -1):
            output = extra_spacing(c, 1, f"{c}", output)

        print(output)
        spaces -= 2

    return None


inverted_numerical_pyramid_pattern(n)
```

```python
def hollow_diagonal_pattern(n=4, pattern="*"):

    space = 1

    # top half pattern
    for a in range(1, n + 1):

        output = ""

        # top-left pattern loop
        for b in range(n + 1 - a, 0, -1):
            output = extra_spacing(b, 1, f"{pattern}", output)

        # top-space pattern loop
        for x in range(1, space + 1):
            output = extra_spacing(x, space, " ", output)

        # top-right pattern loop
        for b in range(1, n + 2 - a):
            output = extra_spacing(b, n + 1 - a, f"{pattern}", output)

        print(output)
        space += 2

    # bottom half pattern
    space -= 2

    for a in range(1, n + 1):

        output = ""

        # bottom-left pattern loop
        for c in range(1, a + 1):
            output = extra_spacing(c, a, f"{pattern}", output)

        # bottom-space pattern loop
        for x in range(1, space + 1):
            output = extra_spacing(x, space, " ", output)

        # bottom-right pattern loop
        for d in range(a, 0, -1):
            output = extra_spacing(d, 1, f"{pattern}", output)

        print(output)
        space -= 2

    return None


hollow_diagonal_pattern(n, symbol)
```

```python
def inverted_pyramid_pattern(n=4, pattern="*"):

    space = (n * 2) - 1

    # top half pattern
    for a in range(1, n + 1):

        output = ""

        # top-left pattern loop
        for b in range(1, a + 1):
            output = extra_spacing(b, a, f"{pattern}", output)

        # top-space loop
        for x in range(1, space + 1):
            output = extra_spacing(x, space, " ", output)

        # top-right pattern loop
        for b in range(1, a + 1):
            output = extra_spacing(b, a, f"{pattern}", output)

        print(output)
        space -= 2

    space = 1

    # bottom half pattern
    for a in range(1, n + 1):

        output = ""

        # bottom-left pattern loop
        for c in range(n, a - 1, -1):
            output = extra_spacing(c, a, f"{pattern}", output)

        # bottom-space loop
        for x in range(1, space + 1):
            output = extra_spacing(x, space, " ", output)

        # bottom-right pattern loop
        for c in range(n, a - 1, -1):
            output = extra_spacing(c, a, f"{pattern}", output)

        print(output)
        space += 2

    return None


inverted_pyramid_pattern(n, symbol)
```

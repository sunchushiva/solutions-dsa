```typescript
const n = 5;
const symbol = "%";


function extraSpacing(current: number, end: number, pattern: string, output: string) {

    if (current !== end) {
        output += pattern + " ";
    } else {
        output += pattern;
    }

    return output;
}
```

```typescript
function binaryPattern(n = 4) {

    let pattern = 1

    for (let a = 1; a <= n; a++) {

        let output = "";
        for (let b = 1; b <= a; b++) {

            if (pattern > 1) {
                pattern = 0;
            }

            output = extraSpacing(b, a, `${pattern}`, output);
            pattern++;

        }

        if (a % 2 !== 0) {
            pattern = 0;
        } else {
            pattern = 1;
        }

        console.log(output)
    }


}

binaryPattern(n)
```

```typescript
function numericalIncrementalPattern(n = 4) {

    let pattern = 1;
    for (let a = 1; a <= n; a++) {

        let output = "";

        for (let b = 1; b <= a; b++) {

            output = extraSpacing(b, a, `${pattern}`, output);
            pattern++;
        }

        console.log(output)
    }

}
numericalIncrementalPattern(n)
```

```typescript
function alphabeticIncrementalPattern(n = 4) {

    let pattern = 65

    for (let a = 1; a <= n; a++) {

        let output = "";

        for (let b = 1; b <= a; b++) {

            if (pattern > 90) {
                pattern = 65;
            }

            output = extraSpacing(b, a, String.fromCharCode(pattern), output);
            pattern++;
        }

        console.log(output);

        pattern = 65
    }
}

alphabeticIncrementalPattern(n)
```

```typescript
function alphabeticIncrementalPatternTwo(n = 4) {

    let pattern = 65

    for (let a = 1; a <= n; a++) {

        let output = "";
        if (pattern > 90) {
            pattern = 65;
        }

        for (let b = 1; b <= a; b++) {
            output = extraSpacing(b, a, String.fromCharCode(pattern), output);
        }

        console.log(output);
        pattern++;
    }

}

alphabeticIncrementalPatternTwo(n)
```

```typescript
function hollowSquarePattern(n = 4, pattern = "*") {

    for (let a = 1; a <= n; a++) {

        let output = "";

        // loops for printing patterns of first and last line 

        if (a === 1 || a === n) {
            for (let b = 1; b <= n; b++) {
                output = extraSpacing(b, n, pattern, output)
            }
        } else {

            for (let c = 1; c <= n; c++) {

                if (c === 1 || c === n) {
                    output = extraSpacing(c, n, pattern, output);
                } else {
                    output = extraSpacing(c, n, " ", output);
                }
            }

        }

        console.log(output)

    }

}

hollowSquarePattern(n, symbol)
```

```python
n = 5
symbol = "^"
diff = 4


def extra_spacing(variable, end, pattern, output):

    if variable != end:
        output += pattern + " "
    else:
        output += pattern

    return output
```

```python
def rectangle_pattern(n=4, pattern="*"):

    rectangle = n + diff

    for a in range(1, n + 1):
        output = ""

        for b in range(1, rectangle):
            output = extra_spacing(b, rectangle - 1, pattern, output)

        print(output)

    return None


rectangle_pattern(n, symbol)
```

```python
def hollow_rectangle_pattern(n=4, pattern="*"):

    rectangle = n + diff

    for a in range(1, n + 1):

        output = ""

        if a == 1 or a == n:

            for b in range(1, rectangle):
                output = extra_spacing(b, rectangle - 1, pattern, output)

        else:

            for c in range(1, rectangle):

                if c == 1 or c == rectangle - 1:
                    output = extra_spacing(c, rectangle - 1, pattern, output)
                else:
                    output = extra_spacing(c, rectangle - 1, " ", output)

        print(output)

    return None


hollow_rectangle_pattern(n, symbol)
```

```python
def parrallelogram_pattern(n=4, pattern="4"):

    rectangle = n + diff

    for a in range(1, n + 1):

        output = ""

        for x in range(1, a):
            output = extra_spacing(x, a, " ", output)

        for b in range(1, rectangle + 1):
            output = extra_spacing(b, rectangle, pattern, output)

        print(output)

    return None


parrallelogram_pattern(n, symbol)
```

```python
def hollow_pyramid_pattern(n=4, pattern="*"):

    stars = 1
    initial_space = (n * 2) // 2

    for a in range(1, n + 1):

        output = ""

        for x in range(1, initial_space):
            output = extra_spacing(x, initial_space + 1, " ", output)

        for b in range(1, stars + 1):
            if a == 1 or a == n or b == 1 or b == stars:
                output = extra_spacing(b, stars, pattern, output)
            else:
                output = extra_spacing(x, initial_space + 1, " ", output)

        print(output)
        stars += 2
        initial_space -= 1

    return None


hollow_pyramid_pattern(n, symbol)
```

```python
def inverted_pyramid_pattern(n=4, pattern="*"):

    stars = n * 2
    initial_space = 1

    for a in range(1, n + 1):

        output = ""

        for x in range(1, initial_space):
            output = extra_spacing(x, initial_space, " ", output)

        for b in range(1, stars):
            output = extra_spacing(b, stars - 1, pattern, output)

        print(output)
        stars -= 2
        initial_space += 1

    return None


inverted_pyramid_pattern(n, symbol)
```

```python
def hollow_inverted_pyramid_pattern(n=4, pattern="*"):

    initial_space = 1
    stars = n * 2

    for a in range(1, n + 1):

        output = ""

        for x in range(1, initial_space):
            output = extra_spacing(x, initial_space, " ", output)

        for b in range(1, stars):
            if b == 1 or b == stars - 1 or a == 1 or a == n:
                output = extra_spacing(b, stars - 1, pattern, output)
            else:
                output = extra_spacing(x, initial_space, " ", output)

        print(output)
        initial_space += 1
        stars -= 2

    return None


hollow_inverted_pyramid_pattern(n, symbol)
```

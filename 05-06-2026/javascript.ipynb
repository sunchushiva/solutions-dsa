```typescript
const n = 5
const symbol = "^"

function extraSpacing(variable: number, end: number = n, print: string = symbol, output: string) {
    if (variable != end) {
        output += print + " "
    } else {
        output += print
    }
    return output

}
```

```typescript
function squarePattern(n = 4, pattern = "*") {

    for (let a = 1; a <= n; a++) {

        let output = "";

        for (let b = 1; b <= n; b++) {
            output = extraSpacing(b, n, pattern, output)
        }

        console.log(output)
    }
}

squarePattern(n, symbol)
```

```typescript
function hollowSquarePattern(n = 4, pattern = "*") {

    for (let a = 1; a <= n; a++) {

        let output = "";

        if (a == 1 || a == n) {
            for (let b = 1; b <= n; b++) {
                output = extraSpacing(b, n, pattern, output)
            }
        } else {

            for (let c = 1; c <= n; c++) {

                if (c == 1 || c == n) {
                    output = extraSpacing(c, n, pattern, output)
                } else {
                    output = extraSpacing(c, n, " ", output)
                }
            }

        }

        console.log(output)
    }

}

hollowSquarePattern(n, symbol)
```

```typescript
function rhombusPattern(n = 4, pattern = "*") {

    for (let a = 1; a <= n; a++) {

        let output = "";

        // intial space adding loop for rhombus patterm
        for (let x = 1; x <= a - 1; x++) {
            output = extraSpacing(x, a + 1, " ", output)
        }

        for (let b = 1; b <= n; b++) {

            output = extraSpacing(b, n, pattern, output)

        }

        console.log(output)

    }

}

rhombusPattern(n, symbol)
```

```typescript
function mirroredRhombusPattern(n = 4, pattern = "*") {

    let intialSpace = n - 1

    for (let a = 1; a <= n; a++) {

        let output = ""

        for (let x = 1; x <= intialSpace; x++) {
            output = extraSpacing(x, intialSpace + 1, " ", output)
        }


        for (let b = 1; b <= n; b++) {
            output = extraSpacing(b, n, pattern, output)
        }

        console.log(output)
        intialSpace--;

    }

}

mirroredRhombusPattern(n, symbol)
```

```typescript
function trianglePattern(n = 4, pattern = "*") {

    for (let a = 1; a <= n; a++) {

        let output = ""

        for (let b = 1; b <= a; b++) {
            output = extraSpacing(b, a, pattern, output)
        }

        console.log(output)

    }

}

trianglePattern(n, symbol)
```

```typescript
function pyramindPattern(n = 4, pattern = "*") {

    let stars = 1;
    let intialSpace = (n * 2) - 1
    intialSpace = intialSpace / 2

    for (let a = 1; a <= n; a++) {
        let output = ""

        for (let x = 1; x <= intialSpace; x++) {
            output = extraSpacing(x, intialSpace + 1, " ", output)
        }


        for (let b = 1; b <= stars; b++) {
            output = extraSpacing(b, stars, pattern, output)
        }

        console.log(output)
        stars += 2
        intialSpace -= 1
    }
}

pyramindPattern(n, symbol)
```
